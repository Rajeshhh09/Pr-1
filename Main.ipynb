Task 1.4: Train / test split

Task 1.5: Apply StandardScaler

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_sc = scaler.fit_transform(X_train)
X_test_sc = scaler.transform(X_test)

import numpy as np

print("X_train_sc mean (rounded):", np.round(X_train_sc.mean(axis=0), 2))
print("X_train_sc std (rounded):", np.round(X_train_sc.std(axis=0), 2))
```

# Task 4: Early Stopping

Task 4.1: Build MLP for Early Stopping experiment

```python
model_es_mlp = keras.Sequential([
    keras.layers.Dense(128, activation='relu', input_shape=(30,)),
    keras.layers.Dense(64, activation='relu'),
    keras.layers.Dense(1, activation='sigmoid')
])

model_es_mlp.compile(
    optimizer='adam', 
    loss='binary_crossentropy', 
    metrics=['accuracy']
)
```

Task 4.2: Add EarlyStopping callback

```python
from tensorflow.keras.callbacks import EarlyStopping

es = EarlyStopping(
    monitor='val_loss', 
    patience=15, 
    restore_best_weights=True, 
    verbose=1
)

history_es = model_es_mlp.fit(
    X_train_sc, y_train, 
    epochs=300, 
    batch_size=32, 
    validation_split=0.1, 
    callbacks=[es], 
    verbose=1
)
```

Task 4.4: Plot Early Stopping curves

```python
model_no_es = keras.Sequential([
    keras.layers.Dense(128, activation='relu', input_shape=(30,)),
    keras.layers.Dense(64, activation='relu'),
    keras.layers.Dense(1, activation='sigmoid')
])

model_no_es.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

history_no_es = model_no_es.fit(
    X_train_sc, y_train, 
    epochs=300, 
    batch_size=32, 
    validation_split=0.1, 
    verbose=0
)

plt.figure(figsize=(10, 5))
plt.plot(history_es.history['val_loss'], label='With Early Stopping', color='green', linewidth=2)
plt.plot(history_no_es.history['val_loss'], label='Without Early Stopping (300 Epochs)', color='red', linewidth=2, linestyle='--')
plt.title('Validation Loss Comparison: Early Stopping vs Uncontrolled Training')
plt.xlabel('Epochs')
plt.ylabel('Validation Loss')
plt.legend()
plt.show()
```

# Task 5: Dropout Layers

Task 5.1: Build MLP with Dropout

```python
model_drop = keras.Sequential([
    keras.layers.Dense(128, activation='relu', input_shape=(30,)),
    keras.layers.Dropout(0.3),
    keras.layers.Dense(64, activation='relu'),
    keras.layers.Dropout(0.3),
    keras.layers.Dense(1, activation='sigmoid')
])

model_drop.summary()
```

Task 5.3: Train with Dropout + Early Stopping

```python
model_drop.compile(
    optimizer='adam', 
    loss='binary_crossentropy', 
    metrics=['accuracy']
)

es_dropout = EarlyStopping(
    monitor='val_loss', 
    patience=20, 
    restore_best_weights=True
)

history_drop = model_drop.fit(
    X_train_sc, y_train, 
    epochs=300, 
    batch_size=32, 
    validation_split=0.1, 
    callbacks=[es_dropout], 
    verbose=1
)
```

Task 5.4: Effect of Dropout rate

```python
import matplotlib.pyplot as plt

def build_dropout_model(rate):
    model = keras.Sequential([
        keras.layers.Dense(128, activation='relu', input_shape=(30,)),
        keras.layers.Dropout(rate),
        keras.layers.Dense(64, activation='relu'),
        keras.layers.Dropout(rate),
        keras.layers.Dense(1, activation='sigmoid')
    ])
    model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
    return model

rates = [0.1, 0.3, 0.5]
val_acc_histories = {}

for r in rates:
    m = build_dropout_model(r)
    hist = m.fit(
        X_train_sc, y_train,
        epochs=200,
        batch_size=32,
        validation_split=0.1,
        callbacks=[EarlyStopping(monitor='val_loss', patience=15, restore_best_weights=True)],
        verbose=0
    )
    val_acc_histories[r] = hist.history['val_accuracy']

plt.figure(figsize=(10, 5))
for r in rates:
    plt.plot(val_acc_histories[r], label=f'Dropout Rate: {r}', linewidth=2)
plt.title('Dropout Rate Comparison — Validation Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Validation Accuracy')
plt.legend()
plt.show()
```

Task 5.5: Evaluate Dropout model

```python
drop_loss, drop_acc = model_drop.evaluate(X_test_sc, y_test, verbose=0)
print(f"Test Accuracy with Dropout (0.3): {drop_acc:.4f}")
```

# Task 6: Regularization (L1, L2, L1-L2)

Task 6.1: Build MLP with L2 Regularization

```python
from tensorflow.keras import regularizers

model_l2 = keras.Sequential([
    keras.layers.Dense(128, activation='relu', input_shape=(30,), kernel_regularizer=regularizers.l2(0.001)),
    keras.layers.Dense(64, activation='relu', kernel_regularizer=regularizers.l2(0.001)),
    keras.layers.Dense(1, activation='sigmoid')
])

model_l2.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

history_l2 = model_l2.fit(
    X_train_sc, y_train,
    epochs=300,
    batch_size=32,
    validation_split=0.1,
    callbacks=[EarlyStopping(monitor='val_loss', patience=20, restore_best_weights=True)],
    verbose=1
)
```

Task 6.2: WHY L2 Regularization — MarkdownNote: L2 regularization adds a penalty proportional to the sum of squared weights ($\lambda \sum w^2$) to the loss function at each training step. This discourages excessively large weight magnitudes, gently pushing all weights toward zero without driving them completely to zero. The result is a smoother, more stable decision boundary where the network relies on a distributed set of small weights instead of over-relying on a few dominant features, which lowers variance and curbs overfitting. A penalty factor of $\lambda = 0.001$ provides a moderate regularizing effect; an excessively high $\lambda$ would over-shrink weights and lead to underfitting.

Task 6.3: Build MLP with L1 Regularization

```python
model_l1l2 = keras.Sequential([
    keras.layers.Dense(128, activation='relu', input_shape=(30,), kernel_regularizer=regularizers.l1_l2(l1=0.0001, l2=0.001)),
    keras.layers.Dense(64, activation='relu', kernel_regularizer=regularizers.l1_l2(l1=0.0001, l2=0.001)),
    keras.layers.Dense(1, activation='sigmoid')
])

model_l1l2.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

history_l1l2 = model_l1l2.fit(
    X_train_sc, y_train,
    epochs=100,
    batch_size=32,
    validation_split=0.1,
    callbacks=[EarlyStopping(monitor='val_loss', patience=20, restore_best_weights=True)],
    verbose=0
)
```

Task 7.1: Best model: combine Dropout + L2 + Early Stopping

```python
model_final = keras.Sequential([
    keras.layers.Dense(128, activation='relu', input_shape=(30,), kernel_regularizer=regularizers.l2(0.001)),
    keras.layers.Dropout(0.3),
    keras.layers.Dense(64, activation='relu', kernel_regularizer=regularizers.l2(0.001)),
    keras.layers.Dropout(0.3),
    keras.layers.Dense(1, activation='sigmoid')
])

model_final.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

history_final = model_final.fit(
    X_train_sc, y_train,
    epochs=300,
    batch_size=32,
    validation_split=0.1,
    callbacks=[EarlyStopping(monitor='val_loss', patience=20, restore_best_weights=True)],
    verbose=1
)
```

Task 7.3: Business insight report — Markdown

Clinical Recommendation & Trade-offs: For deployment in a medical decision-support tool, the combined regularized model (Dropout + L2 + Early Stopping) is strongly recommended due to its high generalisation capacity and robustness against overfitting. In cancer diagnosis, minimizing False Negatives (missing a malignant case, predicting benign when it is actually malignant) is critically more important than False Positives, as a missed diagnosis delays life-saving treatment. Therefore, high Recall is prioritised.

Decision Threshold Tuning: The default classification threshold is 0.5. For a high-stakes clinical screening tool, this threshold should be lowered (e.g., to 0.3 or 0.35). Lowering the threshold increases sensitivity (Recall), ensuring fewer malignant cases slip through as false negatives, at the slight cost of an increased false positive rate which can be cleared via secondary clinical reviews.

Impact of Techniques: Among all techniques, Early Stopping combined with Dropout had the largest positive impact on generalisation. Early stopping successfully prevented the network from memorizing training noise during extended training epochs, while Dropout effectively curbed co-adaptation among high-capacity hidden layers.
